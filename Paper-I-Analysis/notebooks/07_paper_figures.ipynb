# 07 — Paper figures: solar‑frame polarisation composites (+ MGN)

For each HDR method (Figs. 6 and 7 of the paper):

1. instrument‑frame triplet (0°, +60°, −60°) → Stokes I, Q, U → triplet with the
   reference polariser along solar North (rotation by −(168° − 26.24°) = −141.76°);
2. colour composite: hue = polarisation fractions, luminance = γ‑stretched total;
3. same with **MGN** (σ = 10, 20, 40, 80 px) applied to the total intensity;
4. two‑panel figure with compass rose and colour wheel; a moon‑masked variant too.

Outputs: `figures/expnorm_hdr_solar_frame.png`, `figures/ldic_hdr_solar_frame.png`
(+ `_moon_masked`).  Legacy source: `exposure_norm_hdr_paper_figure_Paras.ipynb`,
`ldic_hdr_paper_figure_Paras.ipynb`.

```python
import sys, time
sys.path.insert(0, "..")          # config.py / utils.py live one level up
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy.io import fits

import config, utils
%matplotlib inline
```

```python
METHODS = {
    "expnorm": ("Exposure Normalised HDR (Solar Frame)", "Exposure Normalised HDR + MGN (Solar Frame)"),
    "ldic":    ("LDIC HDR (Solar Frame)",                "LDIC HDR + MGN (Solar Frame)"),
}
print(f"solar North angle {config.SOLAR_NORTH_ANGLE_DEG:.2f} deg  ->  frame rotation phi = {config.SOLAR_FRAME_PHI_DEG:.2f} deg")
```

```python
for method, titles in METHODS.items():
    t0 = time.time()
    pol = [fits.getdata(utils.hdr_filename(method, p)).astype(float) for p in config.POLARIZER_POSITIONS]
    hdr_solar = utils.rotate_to_solar_frame(*pol)
    del pol
    hdr_plot = utils.make_hdr_composite(hdr_solar)
    mgn_plot = utils.make_mgn_polarimetric_composite(hdr_solar)
    del hdr_solar
    utils.plot_solar_frame_panels(hdr_plot, mgn_plot, titles, config.FIGURES_DIR / f"{method}_hdr_solar_frame.png")
    utils.plot_solar_frame_panels(hdr_plot, mgn_plot, titles, config.FIGURES_DIR / f"{method}_hdr_solar_frame_moon_masked.png", moon_mask=True)
    plt.show()
    print(f"{method}: done in {time.time()-t0:.0f} s")
```
